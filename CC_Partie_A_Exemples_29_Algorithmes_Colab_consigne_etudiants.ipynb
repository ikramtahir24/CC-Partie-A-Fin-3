```python
# ==============================================================================
# Q7. CODE AUTONOME POUR LES EXPÉRIMENTATIONS (PROTOCOLES PAR FAMILLE)
# ==============================================================================

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, TimeSeriesSplit
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import Ridge
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, adjusted_rand_score, f1_score
import gymnasium as gym

print("=== EXÉCUTION DE LA Q7 (PROTOCOLES EXPÉRIMENTAUX) ===\n")

# ------------------------------------------------------------------------------
# 1. CAS SUPERVISÉ : Séparation stricte Train / Validation / Test
# ------------------------------------------------------------------------------
print("--- 1. EXPÉRIMENTATION SUPERVISÉE (Train / Validation / Test) ---")
from sklearn.datasets import load_wine
wine = load_wine()
X_sup, y_sup = wine.data, wine.target

# Séparation en 3 ensembles : 60% Train, 20% Validation, 20% Test
X_temp, X_test, y_temp, y_test = train_test_split(X_sup, y_sup, test_size=0.20, random_state=42, stratify=y_sup)
X_train, X_val, y_train, y_val = train_test_split(X_temp, y_temp, test_size=0.25, random_state=42, stratify=y_temp)

scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)  # Ajustement UNIQUEMENT sur le train (pas de data leakage)
X_val_s = scaler.transform(X_val)
X_test_s = scaler.transform(X_test)

clf = RandomForestClassifier(n_estimators=100, random_state=42)
clf.fit(X_train_s, y_train)

val_score = f1_score(y_val, clf.predict(X_val_s), average='weighted')
test_score = f1_score(y_test, clf.predict(X_test_s), average='weighted')

print(f"F1-score sur Validation : {val_score:.4f}")
print(f"F1-score sur Test final : {test_score:.4f}\n")


# ------------------------------------------------------------------------------
# 2. CAS SÉRIES TEMPORELLES : Validation Chronologique (Time Series Split)
# ------------------------------------------------------------------------------
print("--- 2. EXPÉRIMENTATION SÉRIES TEMPORELLES (Validation Chronologique) ---")
from sklearn.datasets import load_diabetes
diabetes = load_diabetes()
X_ts, y_ts = diabetes.data, diabetes.target

tscv = TimeSeriesSplit(n_splits=3)
fold = 1
for train_index, test_index in tscv.split(X_ts):
    X_tr, X_te = X_ts[train_index], X_ts[test_index]
    y_tr, y_te = y_ts[train_index], y_ts[test_index]

    model_ts = Ridge(alpha=1.0).fit(X_tr, y_tr)
    score_r2 = model_ts.score(X_te, y_te)
    print(f"Fold {fold} - Score R² chronologique : {score_r2:.4f}")
    fold += 1
print()


# ------------------------------------------------------------------------------
# 3. CAS NON SUPERVISÉ : Critères Internes et Évaluation Externe
# ------------------------------------------------------------------------------
print("--- 3. EXPÉRIMENTATION NON SUPERVISÉE (Silhouette & ARI) ---")
X_scaled_all = StandardScaler().fit_transform(X_sup)

kmeans = KMeans(n_clusters=3, n_init=10, random_state=42)
labels_pred = kmeans.fit_predict(X_scaled_all)

# Critère interne (Indice de Silhouette)
sil_score = silhouette_score(X_scaled_all, labels_pred)
# Évaluation externe (en comparant avec les vraies étiquettes y_sup disponibles)
ari_score = adjusted_rand_score(y_sup, labels_pred)

print(f"Indice de Silhouette (Critère interne) : {sil_score:.4f}")
print(f"Rand Index Ajusté (Évaluation externe vs vérité) : {ari_score:.4f}\n")


# ------------------------------------------------------------------------------
# 4. CAS APPRENTISSAGE PAR RENFORCEMENT : Environnement Gymnasium (FrozenLake)
# ------------------------------------------------------------------------------
print("--- 4. EXPÉRIMENTATION RENFORCEMENT (Environnement Gymnasium) ---")
# Utilisation d'un environnement public Gymnasium au lieu d'un CSV fixe
env = gym.make('FrozenLake-v1', is_slippery=False)
Q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(42)

for episode in range(1000):
    etat, _ = env.reset(seed=42 + episode)
    epsilon = max(0.05, 1 - episode / 700)

    for pas in range(100):
        action = rng.integers(4) if rng.random() < epsilon else np.argmax(Q_table[etat])
        suivant, recompense, termine, tronque, _ = env.step(action)

        # Mise à jour Q-learning de l'interaction agent-environnement
        cible = recompense + (0 if termine else 0.95 * np.max(Q_table[suivant]))
        Q_table[etat, action] += 0.2 * (cible - Q_table[etat, action])
        etat = suivant

        if termine or tronque:
            break

env.close()
print("Entraînement de l'agent par renforcement sur l'environnement Gymnasium exécuté avec succès.")
```

```python
# ==============================================================================
# Q8. CODE AUTONOME ET SÉCURISÉ POUR LES TABLEAUX COMPARATIFS (FAMILLES 1 À 5)
# ==============================================================================

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error, mean_squared_error, r2_score,
    precision_score, recall_score, f1_score, roc_auc_score,
    silhouette_score, adjusted_rand_score
)

print("=== EXÉCUTION DE LA Q8 (COMPARAISONS PAR FAMILLE) ===\n")

# ------------------------------------------------------------------------------
# 1. FAMILLE : RÉGRESSION (Linéaire vs Random Forest)
# ------------------------------------------------------------------------------
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor

data_reg = load_diabetes()
Xr_ent, Xr_test, yr_ent, yr_test = train_test_split(data_reg.data, data_reg.target, test_size=0.25, random_state=42)

scaler_reg = StandardScaler()
Xr_ent_s = scaler_reg.fit_transform(Xr_ent)
Xr_test_s = scaler_reg.transform(Xr_test)

reg_ref = LinearRegression().fit(Xr_ent_s, yr_ent)
pred_ref = reg_ref.predict(Xr_test_s)

reg_adv = RandomForestRegressor(n_estimators=100, random_state=42).fit(Xr_ent, yr_ent)
pred_adv = reg_adv.predict(Xr_test)

tab_reg = pd.DataFrame({
    'Méthode': ['Régression Linéaire (Référence)', 'Random Forest Regressor (Avancé)'],
    'MAE': [mean_absolute_error(yr_test, pred_ref), mean_absolute_error(yr_test, pred_adv)],
    'RMSE': [np.sqrt(mean_squared_error(yr_test, pred_ref)), np.sqrt(mean_squared_error(yr_test, pred_adv))],
    'R²': [r2_score(yr_test, pred_ref), r2_score(yr_test, pred_adv)]
})

print("--- 1. TABLEAU COMPARATIF : RÉGRESSION ---")
display(tab_reg)


# ------------------------------------------------------------------------------
# 2. FAMILLE : CLASSIFICATION (Régression Logistique vs XGBoost)
# ------------------------------------------------------------------------------
from sklearn.datasets import load_wine
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier

data_cls = load_wine()
Xc_ent, Xc_test, yc_ent, yc_test = train_test_split(data_cls.data, data_cls.target, test_size=0.25, random_state=42, stratify=data_cls.target)

scaler_cls = StandardScaler()
Xc_ent_s = scaler_cls.fit_transform(Xc_ent)
Xc_test_s = scaler_cls.transform(Xc_test)

# Forcer les cibles en tableaux 1D stricts
yc_ent_1d = np.ravel(yc_ent)
yc_test_1d = np.ravel(yc_test)

cls_ref = LogisticRegression(random_state=42, max_iter=1000).fit(Xc_ent_s, yc_ent_1d)
p_ref = cls_ref.predict(Xc_test_s)
proba_ref = cls_ref.predict_proba(Xc_test_s)

cls_adv = XGBClassifier(n_estimators=80, max_depth=3, random_state=42, eval_metric='logloss').fit(Xc_ent, yc_ent_1d)
p_adv = cls_adv.predict(Xc_test)
proba_adv = cls_adv.predict_proba(Xc_test)

tab_cls = pd.DataFrame({
    'Méthode': ['Régression Logistique (Référence)', 'XGBoost Classifier (Avancé)'],
    'Précision': [precision_score(yc_test_1d, p_ref, average='weighted'), precision_score(yc_test_1d, p_adv, average='weighted')],
    'Rappel': [recall_score(yc_test_1d, p_ref, average='weighted'), recall_score(yc_test_1d, p_adv, average='weighted')],
    'F1-score': [f1_score(yc_test_1d, p_ref, average='weighted'), f1_score(yc_test_1d, p_adv, average='weighted')],
    'ROC-AUC': [roc_auc_score(yc_test_1d, proba_ref, multi_class='ovr'), roc_auc_score(yc_test_1d, proba_adv, multi_class='ovr')]
})

print("\n--- 2. TABLEAU COMPARATIF : CLASSIFICATION ---")
display(tab_cls)


# ------------------------------------------------------------------------------
# 3. FAMILLE : CLUSTERING (K-Means vs DBSCAN)
# ------------------------------------------------------------------------------
from sklearn.cluster import KMeans, DBSCAN

X_scaled = StandardScaler().fit_transform(data_cls.data)
kmeans = KMeans(n_clusters=3, n_init=10, random_state=42).fit(X_scaled)
dbscan = DBSCAN(eps=1.8, min_samples=5).fit(X_scaled)

sil_db = silhouette_score(X_scaled, dbscan.labels_) if len(set(dbscan.labels_)) > 1 else -1

tab_clu = pd.DataFrame({
    'Algorithme': ['K-Means (Référence)', 'DBSCAN (Avancé)'],
    'Indice de Silhouette': [silhouette_score(X_scaled, kmeans.labels_), sil_db],
    'Rand Index Ajusté (vs Vérité)': [adjusted_rand_score(data_cls.target, kmeans.labels_), adjusted_rand_score(data_cls.target, dbscan.labels_)],
    'Clusters détectés': [len(set(kmeans.labels_)), len(set(dbscan.labels_)) - (1 if -1 in dbscan.labels_ else 0)]
})

print("\n--- 3. TABLEAU COMPARATIF : CLUSTERING ---")
display(tab_clu)


# ------------------------------------------------------------------------------
# 4. FAMILLE : RÉDUCTION DE DIMENSION (ACP vs UMAP)
# ------------------------------------------------------------------------------
from sklearn.decomposition import PCA
import umap

pca = PCA(n_components=2).fit(X_scaled)
var_exp = pca.explained_variance_ratio_.sum()

reducer = umap.UMAP(n_components=2, random_state=42)
_ = reducer.fit_transform(X_scaled)

tab_dim = pd.DataFrame({
    'Méthode': ['ACP / PCA (Référence)', 'UMAP (Avancé)'],
    'Type de projection': ['Linéaire globale', 'Non linéaire (Voisinage local)'],
    'Métrique d’évaluation': [f"Variance expliquée: {var_exp:.2%}", "Préservation topologique des distances"]
})

print("\n--- 4. TABLEAU COMPARATIF : RÉDUCTION DE DIMENSION ---")
display(tab_dim)


# ------------------------------------------------------------------------------
# 5. FAMILLE : RENFORCEMENT (Q-Learning vs SARSA)
# ------------------------------------------------------------------------------
import gymnasium as gym

def simuler_rl(algo='q_learning', episodes=1000):
    env = gym.make('FrozenLake-v1', is_slippery=False)
    Q = np.zeros((env.observation_space.n, env.action_space.n))
    rng = np.random.default_rng(42)
    succes = []

    for ep in range(episodes):
        etat, _ = env.reset(seed=42 + ep)
        epsilon = max(0.05, 1 - ep / (episodes * 0.7))
        total_r = 0
        action = rng.integers(4) if rng.random() < epsilon else np.argmax(Q[etat])

        for _ in range(100):
            suivant, r, termine, tronque, _ = env.step(action)
            total_r += r
            action_suiv = rng.integers(4) if rng.random() < epsilon else np.argmax(Q[suivant])

            if algo == 'q_learning':
                cible = r + (0 if termine else 0.95 * np.max(Q[suivant]))
            else:
                cible = r + (0 if termine else 0.95 * Q[suivant, action_suiv])

            Q[etat, action] += 0.2 * (cible - Q[etat, action])
            etat, action = suivant, action_suiv
            if termine or tronque:
                break
        succes.append(total_r)
    env.close()
    return np.mean(succes[-100:])

tab_rl = pd.DataFrame({
    'Algorithme': ['Q-Learning (Off-Policy)', 'SARSA (On-Policy)'],
    'Environnement': ['FrozenLake-v1', 'FrozenLake-v1'],
    'Taux de réussite final (100 derniers épisodes)': [f"{simuler_rl('q_learning'):.2%}", f"{simuler_rl('sarsa'):.2%}"]
})

print("\n--- 5. TABLEAU COMPARATIF : APPRENTISSAGE PAR RENFORCEMENT ---")
display(tab_rl)
```

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```
